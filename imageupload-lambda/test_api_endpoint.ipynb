```python
import requests, io
from PIL import Image
```

```python
API_ENDPOINT = 'https://kb7e5zmp4m.execute-api.us-east-1.amazonaws.com/dev'
image_path = 'cfu_positive.jpg'
```

```python
im = Image.open(image_path)
# Create in-memory JPEG
buffer = io.BytesIO()
im.save(buffer, format="JPEG")

# Make request to the endpoint
r = requests.post(url=API_ENDPOINT, data=buffer.getvalue())
# Here is the request
Image.open(io.BytesIO(r.content))
```
